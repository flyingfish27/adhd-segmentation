```python
import os

files = os.listdir("data")

f_ids = sorted([f.replace("_F.csv", "") for f in files if f.endswith("_F.csv")])
t_ids = sorted([f.replace("_T.csv", "") for f in files if f.endswith("_T.csv")])

print("F文件:", len(f_ids), f_ids)
print("T文件:", len(t_ids), t_ids)
```

```python
label_ids = set(labels["ID"])
data_ids = set(f_ids) | set(t_ids)

both = data_ids & label_ids
data_only = data_ids - label_ids
label_only = label_ids - data_ids

print("有数据+有标签:", len(both), sorted(both))
print("有数据无标签:", sorted(data_only))
print("有标签无数据:", sorted(label_only))
```

```python
labels["ID"] = labels["ID"].replace("Q27", "Z27")

sdq_cols = [c for c in labels.columns if c.startswith("SDQ")]
snap_cols = [c for c in labels.columns if c.startswith("a")]

labels["sdq_missing"] = labels[sdq_cols].isna().sum(axis=1)
labels["snap_missing"] = labels[snap_cols].isna().sum(axis=1)

problem = labels[(labels["sdq_missing"] > 0) | (labels["snap_missing"] > 0)]
print(problem[["ID", "sdq_missing", "snap_missing"]])
```

```python
sdq_hyper_items = ["SDQ2", "SDQ10", "SDQ15", "SDQ21", "SDQ25"]

s = labels[sdq_hyper_items].copy()
s["SDQ21"] = 4 - s["SDQ21"]
s["SDQ25"] = 4 - s["SDQ25"]
labels["sdq_hyper"] = (s - 1).sum(axis=1, min_count=5)

labels["snap_inatt"] = labels[[f"a{i}" for i in range(1, 10)]].mean(axis=1)
labels["snap_hyper"] = labels[[f"a{i}" for i in range(10, 19)]].mean(axis=1)

labels[["ID", "sdq_hyper", "snap_inatt", "snap_hyper"]].sort_values("sdq_hyper", ascending=False).head(15)
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))

axes[0].hist(labels["sdq_hyper"].dropna(), bins=range(0, 12), edgecolor="black")
axes[0].set_title("SDQ Hyperactivity (0-10)")
axes[0].set_xlabel("score"); axes[0].set_ylabel("children")

axes[1].hist(labels["snap_inatt"].dropna(), bins=12, edgecolor="black")
axes[1].set_title("SNAP-IV Inattention (mean, 1-4 scale)")
axes[1].set_xlabel("score")

axes[2].hist(labels["snap_hyper"].dropna(), bins=12, edgecolor="black")
axes[2].set_title("SNAP-IV Hyperactivity (mean, 1-4 scale)")
axes[2].set_xlabel("score")

plt.tight_layout()
plt.show()
```

```python
labels["snap_inatt_sum"] = labels[[f"a{i}" for i in range(1, 10)]].sum(axis=1, min_count=9)
labels["snap_hyper_sum"] = labels[[f"a{i}" for i in range(10, 19)]].sum(axis=1, min_count=9)

result = labels[["ID", "snap_inatt_sum", "snap_hyper_sum"]].sort_values(
    "snap_inatt_sum", ascending=False
)
print(result.to_string())
```

```python
inatt_rank = labels[["ID", "snap_inatt_sum"]].dropna().sort_values(
    "snap_inatt_sum", ascending=False
)
print("=== SNAP 注意缺陷维度（a1-a9总分，满分36）===")
print(inatt_rank.to_string(index=False))

hyper_rank = labels[["ID", "snap_hyper_sum"]].dropna().sort_values(
    "snap_hyper_sum", ascending=False
)
print("\n=== SNAP 多动/冲动维度（a10-a18总分，满分36）===")
print(hyper_rank.to_string(index=False))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ← 改成你放 H44_F.csv / H44_T.csv 的文件夹路径
BASE = './data'   # 例如 './data' 或 'C:/Users/你/Desktop/adhd'

# 我们只需要这 7 列:时间戳 + 含重力三轴 + 去重力三轴
cols = ['loggingTime(txt)',
        'accelerometerAccelerationX(G)', 'accelerometerAccelerationY(G)', 'accelerometerAccelerationZ(G)',
        'motionUserAccelerationX(G)',    'motionUserAccelerationY(G)',    'motionUserAccelerationZ(G)']

def load(path, sep):
    # sep: _F 传 ',' , _T 传 ';'  ← 两个文件分隔符不同,这是最容易踩的坑
    df = pd.read_csv(f'{BASE}/{path}', sep=sep, usecols=cols)

    # 把文本时间戳解析成时间对象,再算"从0开始的相对秒数"当横轴
    t = pd.to_datetime(df['loggingTime(txt)'], format='ISO8601')
    df['elapsed_s'] = (t - t.iloc[0]).dt.total_seconds()

    # 含重力:三轴平方和开根 = 合成幅值
    df['acc_mag'] = np.sqrt(df['accelerometerAccelerationX(G)']**2
                          + df['accelerometerAccelerationY(G)']**2
                          + df['accelerometerAccelerationZ(G)']**2)
    # 去重力:同样算幅值
    df['usr_mag'] = np.sqrt(df['motionUserAccelerationX(G)']**2
                          + df['motionUserAccelerationY(G)']**2
                          + df['motionUserAccelerationZ(G)']**2)
    return df

F = load('H44_F.csv', ',')   # 1分钟 100Hz
T = load('H44_T.csv', ';')   # 1小时 29Hz  ← 注意分号

# 一个小工具:在给定的子图 ax_ 上画 x/y/z/幅值 4 条线
# prefix: 含重力传 'accelerometerAcceleration', 去重力传 'motionUserAcceleration'
# xcol:   横轴用哪列   mag: 幅值用哪列
def draw(prefix, df, xcol, xlabel, mag, ax_):
    ax_.plot(df[xcol], df[prefix+'X(G)'], lw=0.5, label='x')
    ax_.plot(df[xcol], df[prefix+'Y(G)'], lw=0.5, label='y')
    ax_.plot(df[xcol], df[prefix+'Z(G)'], lw=0.5, label='z')
    ax_.plot(df[xcol], df[mag], lw=0.7, color='black', label='magnitude')
    ax_.set_xlabel(xlabel); ax_.set_ylabel('acceleration (G)')
    ax_.legend(loc='upper right', fontsize=8); ax_.grid(alpha=0.3)

# 2行2列的画布
fig, ax = plt.subplots(2, 2, figsize=(16, 9))

# _T 横轴改用分钟,一小时看得舒服点
T['elapsed_min'] = T['elapsed_s'] / 60

# 左上 _F 含重力 / 右上 _F 去重力 / 左下 _T 含重力 / 右下 _T 去重力
draw('accelerometerAcceleration', F, 'elapsed_s',   'time (s)',   'acc_mag', ax[0,0])
ax[0,0].set_title('H44 _F (1min,100Hz) — WITH gravity')
draw('motionUserAcceleration',    F, 'elapsed_s',   'time (s)',   'usr_mag', ax[0,1])
ax[0,1].set_title('H44 _F (1min,100Hz) — NO gravity')
draw('accelerometerAcceleration', T, 'elapsed_min', 'time (min)', 'acc_mag', ax[1,0])
ax[1,0].set_title('H44 _T (1hour,29Hz) — WITH gravity')
draw('motionUserAcceleration',    T, 'elapsed_min', 'time (min)', 'usr_mag', ax[1,1])
ax[1,1].set_title('H44 _T (1hour,29Hz) — NO gravity')

plt.tight_layout()
plt.show()

# 顺便打印几个均值,帮我们读图
print('含重力幅值  _F=%.3f  _T=%.3f' % (F['acc_mag'].mean(), T['acc_mag'].mean()))
print('去重力幅值  _F=%.3f  _T=%.3f' % (F['usr_mag'].mean(), T['usr_mag'].mean()))
```

```python
import glob
import pandas as pd
from collections import Counter

# 自动找出 data 文件夹里所有 _T 文件(不用手写文件名)
t_files = sorted(glob.glob('./data/*_T.csv'))
print('找到', len(t_files), '个 _T 文件')   # 先确认找到了几个,应该接近全部被试数

# 用 Counter 把每个文件里 activity 列的计数累加起来
total = Counter()

for path in t_files:
    # 只读 activity(txt) 这一列; _T 是分号分隔
    col = pd.read_csv(path, sep=';', usecols=['activity(txt)'])['activity(txt)']
    # value_counts() 数每个取值出现多少次,再累加到总表
    total.update(col.value_counts().to_dict())

# 打印:所有出现过的类别 + 各自总次数(按次数从多到少)
print('\n=== 所有 _T 文件里 activity(txt) 出现过的类别 ===')
for label, count in total.most_common():
    print(f'{label:<15} {count:>12,} 次')
```

```python
import glob
import pandas as pd
from collections import Counter

t_files = sorted(glob.glob('./data/*_T.csv'))
print('找到', len(t_files), '个 _T 文件\n')

total = Counter()
bad_files = []   # 记录读不出 activity 列的文件

for path in t_files:
    try:
        # 还是先按分号读
        col = pd.read_csv(path, sep=';', usecols=['activity(txt)'])['activity(txt)']
        total.update(col.value_counts().to_dict())
    except ValueError:
        # 读不出来 -> 把这个文件记下来,并看看它实际有哪些列(只读前1行看列名)
        peek = pd.read_csv(path, sep=';', nrows=1)
        bad_files.append((path, list(peek.columns)))

print('=== 正常读到的类别 ===')
for label, count in total.most_common():
    print(f'{label:<15} {count:>12,} 次')

print(f'\n=== 有问题的文件:{len(bad_files)} 个 ===')
for path, cols in bad_files:
    print('\n文件:', path)
    print('它的列数:', len(cols))
    print('前几个列名:', cols[:3])   # 只看前3个,判断分隔符对不对
```

```python
# --- Cell 3: 第一次读入 H45_F.csv,看看它长什么样 ---
import pandas as pd
from pathlib import Path

# 拼出这个文件的完整路径:当前目录 / data / 文件名
f_path = Path.cwd() / "data" / "H45_F.csv"

# 用 pandas 读 CSV。
# 关键点:H45_F 是用【逗号】分隔的,所以 sep=","(这其实是默认值,写出来是为了让你看清)
# 之后读 _T 文件时要改成 sep=";"(分号)—— 这个坑我们前面聊过
df_F = pd.read_csv(f_path, sep=",")

# df 是 DataFrame(数据表)的习惯叫法。下面几行都是“看一眼”的操作:

# (1) 形状:多少行、多少列。返回 (行数, 列数)
print("形状 (行数, 列数):", df_F.shape)

# (2) 列名:总共有哪些列。先只看前 20 个列名,免得刷屏
print("\n前 20 个列名:")
print(list(df_F.columns)[:20])

# (3) 前 3 行长什么样(只挑我们最关心的几列看,不然 59 列会横向刷爆屏幕)
cols_we_care = [
    "loggingTime(txt)",
    "accelerometerAccelerationX(G)",
    "accelerometerAccelerationY(G)",
    "accelerometerAccelerationZ(G)",
]
print("\n我们关心的几列,前 3 行:")
print(df_F[cols_we_care].head(3))
```

```python
# --- Cell 4: 计算合成幅值 magnitude,并把时间轴处理成“从0秒开始” ---
import numpy as np   # numpy:做数学运算的库,算平方根、平方很方便

# 为了少打字,把三列加速度取出来,起个短名字
x = df_F["accelerometerAccelerationX(G)"]
y = df_F["accelerometerAccelerationY(G)"]
z = df_F["accelerometerAccelerationZ(G)"]

# 算 magnitude = 根号(x² + y² + z²)
# np.sqrt = 平方根;  x**2 = x 的平方
magnitude = np.sqrt(x**2 + y**2 + z**2)

# 把算好的 magnitude 作为新的一列,加回到表里,方便后面画图
df_F["magnitude"] = magnitude

# 处理时间轴:用加速度计的时间戳(秒),减去第一个值 → 让时间从 0 开始
t_raw = df_F["accelerometerTimestamp_sinceReboot(s)"]
df_F["time_s"] = t_raw - t_raw.iloc[0]   # iloc[0] = 第一行的值

# --- 下面全是“检查一眼”,确认算得对 ---
print("magnitude 前 3 个值:")
print(magnitude.head(3).values)

print("\nmagnitude 的统计(看范围是否合理):")
print("  最小值:", round(magnitude.min(), 3))
print("  最大值:", round(magnitude.max(), 3))
print("  平均值:", round(magnitude.mean(), 3))

print("\n时间轴:第一个值应为 0,最后一个值应约为 68 秒")
print("  time_s 起点:", round(df_F["time_s"].iloc[0], 3))
print("  time_s 终点:", round(df_F["time_s"].iloc[-1], 3))
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# ← 改成你放 H45_F.csv / H45_T.csv 的文件夹路径
BASE = './data'   # 例如 './data' 或 'C:/Users/你/Desktop/adhd'

# 我们只需要这 7 列:时间戳 + 含重力三轴 + 去重力三轴
cols = ['loggingTime(txt)',
        'accelerometerAccelerationX(G)', 'accelerometerAccelerationY(G)', 'accelerometerAccelerationZ(G)',
        'motionUserAccelerationX(G)',    'motionUserAccelerationY(G)',    'motionUserAccelerationZ(G)']

def load(path, sep):
    # sep: _F 传 ',' , _T 传 ';'  ← 两个文件分隔符不同,这是最容易踩的坑
    df = pd.read_csv(f'{BASE}/{path}', sep=sep, usecols=cols)

    # 把文本时间戳解析成时间对象,再算"从0开始的相对秒数"当横轴
    t = pd.to_datetime(df['loggingTime(txt)'], format='ISO8601')
    df['elapsed_s'] = (t - t.iloc[0]).dt.total_seconds()

    # 含重力:三轴平方和开根 = 合成幅值
    df['acc_mag'] = np.sqrt(df['accelerometerAccelerationX(G)']**2
                          + df['accelerometerAccelerationY(G)']**2
                          + df['accelerometerAccelerationZ(G)']**2)
    # 去重力:同样算幅值
    df['usr_mag'] = np.sqrt(df['motionUserAccelerationX(G)']**2
                          + df['motionUserAccelerationY(G)']**2
                          + df['motionUserAccelerationZ(G)']**2)
    return df

F = load('H45_F.csv', ',')   # 1分钟 100Hz
T = load('H45_T.csv', ';')   # 1小时 29Hz  ← 注意分号

# 一个小工具:在给定的子图 ax_ 上画 x/y/z/幅值 4 条线
# prefix: 含重力传 'accelerometerAcceleration', 去重力传 'motionUserAcceleration'
# xcol:   横轴用哪列   mag: 幅值用哪列
def draw(prefix, df, xcol, xlabel, mag, ax_):
    ax_.plot(df[xcol], df[mag], lw=0.7, color='black', label='magnitude')
    ax_.set_xlabel(xlabel); ax_.set_ylabel('acceleration (G)')
    ax_.legend(loc='upper right', fontsize=8); ax_.grid(alpha=0.3)

# 2行2列的画布
fig, ax = plt.subplots(2, 2, figsize=(16, 9))

# _T 横轴改用分钟,一小时看得舒服点
T['elapsed_min'] = T['elapsed_s'] / 60

# 左上 _F 含重力 / 右上 _F 去重力 / 左下 _T 含重力 / 右下 _T 去重力
draw('accelerometerAcceleration', F, 'elapsed_s',   'time (s)',   'acc_mag', ax[0,0])
ax[0,0].set_title('H45 _F (1min,100Hz) — WITH gravity')
draw('motionUserAcceleration',    F, 'elapsed_s',   'time (s)',   'usr_mag', ax[0,1])
ax[0,1].set_title('H45 _F (1min,100Hz) — NO gravity')
draw('accelerometerAcceleration', T, 'elapsed_min', 'time (min)', 'acc_mag', ax[1,0])
ax[1,0].set_title('H45 _T (1hour,29Hz) — WITH gravity')
draw('motionUserAcceleration',    T, 'elapsed_min', 'time (min)', 'usr_mag', ax[1,1])
ax[1,1].set_title('H45 _T (1hour,29Hz) — NO gravity')

plt.tight_layout()
plt.show()

# 顺便打印几个均值,帮我们读图
print('含重力幅值  _F=%.3f  _T=%.3f' % (F['acc_mag'].mean(), T['acc_mag'].mean()))
print('去重力幅值  _F=%.3f  _T=%.3f' % (F['usr_mag'].mean(), T['usr_mag'].mean()))
```
